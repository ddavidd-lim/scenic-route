```python
import osmium
from osmium import filter, osm
import geopandas
```

```python
from pathlib import Path

path = Path(
    "src/data/osm/california-260203.osm.pbf/"
).resolve()

print(path)
```

```python
fp_water = (
    osmium.FileProcessor(path)
    .with_locations()
    .with_filter(filter.EntityFilter(osm.WAY))
    .with_filter(filter.KeyFilter("natural", "water"))
    .with_filter(filter.GeoInterfaceFilter())
)
```

```python
feature_water = geopandas.GeoDataFrame.from_features(
    fp_water, crs="EPSG:4326"
)
```

```python
feature_water.plot()
```
