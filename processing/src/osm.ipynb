```python
import osmium
import geopandas
```

```python
from pathlib import Path

path = Path(
    "src/data/water/california-260203.osm.pbf/"
).resolve()

print(path)
```

```python
fp_water = (
    osmium.FileProcessor(path)
    .with_locations()
    .with_filter(osmium.filter.EntityFilter(osmium.osm.WAY))
    .with_filter(osmium.filter.KeyFilter("natural", "water"))
    .with_filter(osmium.filter.GeoInterfaceFilter())
)
```

```python
feature_water = geopandas.GeoDataFrame.from_features(
    fp_water, crs="EPSG:4326"
)
```

```python
feature_water.plot()
```
